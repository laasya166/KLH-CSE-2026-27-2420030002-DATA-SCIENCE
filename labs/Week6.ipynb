```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
print("First 5 rows of dataset: ")
print(df.head())
mean_values=df.mean()
print("Mean values")
print(mean_values)
median_values=df.median()
print("Median values")
print(median_values)
mode_values=df.mode()
print("Mode values")
print(mode_values)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
print("First 5 rows of dataset: ")
print(df.head())
range_values=df.max()-df.min()
print("Range values:")
print(range_values)
variance_values=df.var
print("Variance values:")
print(variance_values)
std_values=df.std()
print("Standatd deviation values")
print(std_values)
Q1=df.quantile(0.25)
Q3=df.quantile(0.75)
IQR=Q3-Q1
print("Interquartile range")
print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
print("First 5 rows of the dataset")
print(df.head())
skewness_values=df.skew()
print("Skewness values")
print(skewness_values)
kurtosis_values=df.kurt()
print("Kurtosis values")
print(kurtosis_values)
```
